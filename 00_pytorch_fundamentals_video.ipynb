### Matrix multiplication (Convolution)
Two main ways of performing multiplication in neural networks and deep learning

1. Element-wise multiplication (scalar or dot product)
2. Matrix multiplication (cross product or
convolution)

There are two main rules that performing matrix multiplication need to satisfy:

1. The **inner dimensions** must match:

*(3,2) @ (3,2) won't work
*(2,3) @ (3,2) will work.
*(3,2) @ (2,3) will work.

2. The resulting matrix has the shape of the **outer dimensions**
(2,3) @ (3,2) = (2,2)

```python
a = torch.rand(2,3)
print(f"a =  {a} \n")
b = torch.rand(3,2)
print(f"b =  {b}")

print(f" a @ b = {a @ b}")
```

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes an input tensors to a defined shape
* View - return a view of an input tensor of a certain shape, but keep the same memory as the original tensor
* Stacking - combine mulitple tensors on top of each other (vstack - vertical) or side by side (hstack - horizontal)
* Squeeze - removes all 1 dimensions from a tensor
* Unsqueeze  - add all 1 dimensions to a target tensor
* Permute - return a view of the input with dimensions permuted (swapped) in a certain way

```python
# Let's create a tensor
import torch
x = torch.arange(1.,10.)
x, x.shape
```

```python
# Reshape - add an extra dimension <--- aumenta di una dimensione (aggiunge una [])
x_reshaped = x.reshape(1,9) # Deve essere multiplo!
x_reshaped, x_reshaped.shape
```

```python
# Change the view
z = x.view(1,9) # it's quite similar to shape...z is a kind of copy of x
z, z.shape
```

```python
# Changing z changes x (because a view of a tensor shares the same memory as the original input)
z[:,0] = 5 # Cambio tutte gli elementi della prima colonna in 5
z,x
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x,x,x,x],dim = 1) # <--- Li incatasto verticalmente
x_stacked
```

```python
# Squeeze and unsqueeze

# torch.squeeze() - removes all single dimensions from a target tensor
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}\n")

# Remove extra dimensions from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"New tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
# torch.unsqueeze() - add a single dimensions to a target tensor at specific dimension
print(f"Previous target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}\n")

# Add an extra dimensions with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim = 0)
print(f"New tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute - rearranges the dimensions of a target tensor in a spcified order
x_original = torch.rand(size=(224,224,3)) # [height, width, color_channels (RGB)]

# Permute the original tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2,0,1) # Shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}") # Mi switcha l'ordine delle features
```

## PyTorch tensors and NumPy

NumPy is a popular scientific Python numerical computing library. Becuase of this, PyTorch has functionality to interact with it.

* Data in NumPy, want in PyTorch tensor -> torch.from_numpy(ndarray)
* PyTorch tensor -> NumPy -> torch.Tensor.numpy()



```python
# Numpy array to tensor
import torch
import numpy as np


array = np.arange(1.,8)
tensor = torch.from_numpy(array) #WARNING: when converting from numpy -> pytorch, PyTorch reflects numpy's default datatype of float64 unless specified otherwise
#tensor = torch.from_numpy(array).type(torch.float32)

array, tensor #abbiamo un float64 come tipo del tensor, perché lo eredita dall'array NumPy. Questo non è il tipo base del tensor, che è float32. Dobbiamo cambiarlo
```

```python
torch.arange(1.,8.).dtype
```

```python
# Chane the value of array, what will this do to tensor? It doesn't affect the tensor object

array = array + 1
array, tensor
```

```python
# Tensor to NumPy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy() # This will has the tensor datatype (float32)...keep in mind
tensor, numpy_tensor
```

```python
# Change the tensor, what happens to numoy_tensor
tensor = tensor + 1
tensor, numpy_tensor
```

## Reproducibility (trying to take random out of random)

In short, how a neural network learns:

`start with random numbers -> tensor operations -> update random numbers to try and make them better rapresentation of the data -> again -> again -> again`

To reduce the randomness in neural networks and PyTorch comes the concept of a **random seed**.

Essentially, what the random seed does is "flavour" the randomness.

```python
torch.rand(3,3) # questo crea un numero diverso sempre...ma se volessi riproducibilità?
```

```python
import torch
# Create two random tensors

random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B) # Mi torna se due oggetti sono uguali o diversi (= è True ------ ≠ è False)
```

```python
# Let's make some random but reproducible tensors --> uso torch.manual_seed(Random_seed)
import torch

# Set the random seed <--- è come quando setto il seed del TRandom. Il concetto è proprio identico
Random_seed = 42
torch.manual_seed(Random_seed)
random_tensor_C = torch.rand(3,4)

#torch.manual_seed(Random_seed) #<-- Questo riazzera la sequenza pseudocasuale
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
